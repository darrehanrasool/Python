## Question
Calculate population after compound growth.

## Explanation
Enter initial population, annual growth rate as a percent, and time in years.

## How to run
Run the code cell and enter the requested values when prompted.

```python
initial_population = float(input('Enter initial population: '))
annual_growth_percent = float(input('Enter annual growth rate (%): '))
time_years = float(input('Enter time in years: '))

## Compound growth: final population = initial * (1 + rate)^time.
final_population = initial_population * (1 + annual_growth_percent / 100) ** time_years

print(f'Population after {time_years:g} years: {final_population:.2f}')
```
