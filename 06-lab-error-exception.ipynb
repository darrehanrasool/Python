```python
# Practical 6: Error Handling and Exceptions - Handle runtime exceptions including ZeroDivisionError and FileNotFoundError using try-except-else-finally.

def safe_divide(numerator, denominator):
    '''Safely divides two numbers while handling ZeroDivisionError and TypeError.'''
    try:
        result = numerator / denominator
    except ZeroDivisionError:
        return "Error: Cannot divide by zero."
    except TypeError:
        return "Error: Operands must be numeric."
    else:
        return f"Success: {result:.2f}"
    finally:
        pass  # Clean-up actions

def safe_read_file(filepath):
    '''Demonstrates handling FileNotFoundError when accessing disk files.'''
    try:
        with open(filepath, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        return f"Caught Expected Exception: File '{filepath}' does not exist."

print("Division by valid number (10/2):", safe_divide(10, 2))
print("Division by zero (10/0):        ", safe_divide(10, 0))
print("File access test:               ", safe_read_file("non_existent_lab.txt"))
```
