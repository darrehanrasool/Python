```python
# Practical 8: List Comprehensions and Generators - Create lists with comprehensions and yield infinite sequences with generators.
# Part 1: List comprehension - squares of the first 10 natural numbers
squares_of_ten = [x**2 for x in range(1, 11)]
print("Squares of first 10 natural numbers:", squares_of_ten)

# Filtered list comprehension - even numbers squared
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print("Even squares up to 10:", even_squares)

# Part 2: Generator function producing an infinite sequence
def infinite_counter(start=1, step=1):
    '''Yields an unbounded sequence of numbers on demand.'''
    current = start
    while True:
        yield current
        current += step

# Consume first 8 values from the infinite generator safely
counter_gen = infinite_counter(start=100, step=5)
first_values = [next(counter_gen) for _ in range(8)]
print("First 8 values from infinite generator:", first_values)
```
