```python
# Practical 11: Web Scraping - Scrape and extract structured data from HTML pages using BeautifulSoup.
from bs4 import BeautifulSoup

# Standalone structured HTML page representing course faculty and research papers
html_markup = '''
<!DOCTYPE html>
<html>
<head><title>Centre for AI - Research Directory</title></head>
<body>
    <h1>Department Research Publications</h1>
    <div class="paper-card" id="paper-1">
        <h2 class="title">Deep Learning for Natural Language Understanding</h2>
        <p class="author">Author: Dr. Tariq Ahmad</p>
        <span class="citations">Citations: 142</span>
        <a class="link" href="https://doi.org/10.1000/182">Paper DOI Link</a>
    </div>
    <div class="paper-card" id="paper-2">
        <h2 class="title">Reinforcement Learning in Robotics</h2>
        <p class="author">Author: Dar Rehan</p>
        <span class="citations">Citations: 89</span>
        <a class="link" href="https://doi.org/10.1000/183">Paper DOI Link</a>
    </div>
</body>
</html>
'''

soup = BeautifulSoup(html_markup, "html.parser")

page_title = soup.title.string
main_heading = soup.find("h1").get_text()
print(f"Page Title: {page_title}")
print(f"Header:     {main_heading}\n")

# Extract structured cards into Python dictionaries
publications = []
for card in soup.find_all("div", class_="paper-card"):
    title = card.find("h2", class_="title").get_text()
    author = card.find("p", class_="author").get_text().replace("Author: ", "")
    citations = card.find("span", class_="citations").get_text()
    link = card.find("a", class_="link")["href"]
    publications.append({
        "title": title,
        "author": author,
        "citations": citations,
        "url": link
    })

print("Scraped Publications:")
for idx, pub in enumerate(publications, start=1):
    print(f"  {idx}. {pub['title']} | {pub['author']} | {pub['citations']} | {pub['url']}")
```
