```python
# Question 34: Calculate the total bill amount given item unit prices and purchase quantities.
# Formula: Total Bill = Sum of (price * quantity) for all purchased items.

# Realistic grocery bill example with item names, prices, and quantities
item_names = ['Basmati Rice (kg)', 'Fresh Milk (L)', 'Brown Bread (loaf)', 'Eggs (dozen)', 'Sunflower Oil (L)']
prices = [85, 60, 45, 90, 160]
quantities = [2, 3, 1, 2, 1]

total_bill = 0
print(f"{'Item Description':<25} {'Unit Price':<12} {'Quantity':<10} {'Subtotal':<10}")
print('-' * 60)
for i in range(len(prices)):
    subtotal = prices[i] * quantities[i]
    total_bill += subtotal
    print(f"{item_names[i]:<25} Rs.{prices[i]:<9} {quantities[i]:<10} Rs.{subtotal:<10}")

print('-' * 60)
print(f"{'TOTAL BILL AMOUNT:':<49} Rs.{total_bill}")
```
