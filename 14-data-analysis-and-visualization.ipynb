```python
# Practical 14: Data Analysis and Visualization - Analyze datasets with pandas and generate histogram and scatter plot charts using matplotlib.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create synthetic dataset: Student study hours vs exam scores
np.random.seed(42)
hours = np.random.uniform(10, 50, size=50)
noise = np.random.normal(0, 5, size=50)
scores = np.clip(35 + (hours * 1.25) + noise, 0, 100)

df = pd.DataFrame({
    "StudentID": [f"STD_{i:03d}" for i in range(1, 51)],
    "StudyHours": np.round(hours, 1),
    "ExamScore": np.round(scores, 1)
})

# Summary Statistics
print("Dataset Head:")
print(df.head(4).to_string(index=False))
print("\nDescriptive Statistics:")
print(df[["StudyHours", "ExamScore"]].describe().to_string())
corr = df["StudyHours"].corr(df["ExamScore"])
print(f"\nPearson Correlation (StudyHours vs ExamScore): {corr:.4f}")

# Visualization: Histogram and Scatter Plot
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# Plot 1: Histogram
axes[0].hist(df["ExamScore"], bins=8, color="#2b5c8f", edgecolor="black", alpha=0.8)
axes[0].set_title("Distribution of Student Exam Scores")
axes[0].set_xlabel("Exam Score")
axes[0].set_ylabel("Number of Students")
axes[0].grid(axis="y", linestyle="--", alpha=0.6)

# Plot 2: Scatter Plot
axes[1].scatter(df["StudyHours"], df["ExamScore"], color="#d95f02", edgecolor="black", alpha=0.85)
m, b = np.polyfit(df["StudyHours"], df["ExamScore"], 1)
axes[1].plot(df["StudyHours"], m*df["StudyHours"] + b, color="darkred", linestyle="--", label=f"Trend: y={m:.2f}x+{b:.1f}")
axes[1].set_title("Study Hours vs. Exam Score")
axes[1].set_xlabel("Weekly Study Hours")
axes[1].set_ylabel("Final Exam Score")
axes[1].legend()
axes[1].grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.savefig("practical14_visualization.png", dpi=150)
plt.close(fig)
print("Generated and saved visualization figure to 'practical14_visualization.png'.")
```
