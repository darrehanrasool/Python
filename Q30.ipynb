## Question
Design a program that evaluates a user-provided mathematical formula using predefined variables and math functions.

## Explanation
Enter values for x and y, then a Python expression using x, y, and the listed math functions (for example: sqrt(x**2 + y**2) + sin(x)). The expression is evaluated with built-in functions disabled. Only enter a mathematical expression you trust.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

x = float(input('Enter x: '))
y = float(input('Enter y: '))
formula = input('Enter formula using x, y and math functions: ')

## Provide a small namespace of approved math functions and variables.
## eval evaluates the expression in that namespace; built-in functions are disabled.
allowed_names = {
    'x': x, 'y': y,
    'pi': math.pi, 'e': math.e,
    'sin': math.sin, 'cos': math.cos, 'tan': math.tan,
    'sqrt': math.sqrt, 'log': math.log, 'exp': math.exp,
    'abs': abs
}
result = eval(formula, {'__builtins__': {}}, allowed_names)

print(f'Result: {result}')
```
