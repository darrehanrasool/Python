```python
# Practical 12: Web Development - Build a web application with dynamic routes and API endpoints using Flask.
from flask import Flask, jsonify, request

app = Flask(__name__)

# Home route serving HTML greeting
@app.route("/")
def index():
    return "<h1>Welcome to Centre for AI - Python Web Portal</h1>"

# Dynamic parameter route
@app.route("/greet/<name>")
def greet_user(name):
    return f"Hello, {name}! Welcome to the Python Web Framework practical."

# JSON API endpoint
@app.route("/api/info")
def get_info():
    return jsonify({
        "course": "Python Programming Language Lab",
        "course_code": "CAI506C",
        "framework": "Flask",
        "status": "Operational"
    })

# JSON computation endpoint
@app.route("/api/square/<int:number>")
def compute_square(number):
    return jsonify({
        "input": number,
        "square": number ** 2,
        "cube": number ** 3,
        "success": True
    })

# Test endpoints programmatically using Flask test_client
with app.test_client() as client:
    home_res = client.get("/")
    greet_res = client.get("/greet/Rehan")
    api_res = client.get("/api/info")
    calc_res = client.get("/api/square/12")

    print("GET / -> Status:", home_res.status_code, "| Body:", home_res.data.decode()[:40])
    print("GET /greet/Rehan -> Status:", greet_res.status_code, "| Body:", greet_res.data.decode())
    print("GET /api/info -> JSON Response:", api_res.get_json())
    print("GET /api/square/12 -> JSON Response:", calc_res.get_json())
```
