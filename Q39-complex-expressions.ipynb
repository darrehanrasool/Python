### Level 1 — Moderate: Question 1
**Mathematical Expression:**
$$x^2 + 3x + 5$$

**Description:**
Evaluates a standard second-degree quadratic polynomial. For any real number $x$, it computes $x$ squared plus three times $x$, plus five.
* **Domain:** $x \in (-\infty, \infty)$


```python
# Level 1 — Moderate
# Question 1: x² + 3x + 5
import math

x = 2
result = x**2 + 3 * x + 5

print("Expression: x² + 3x + 5")
print(f"Variables:  x = {x}")
print(f"Result:     {result}")
```

### Level 3 — Complex: Question 13
**Mathematical Expression:**
$$\left[\frac{1}{\sqrt{2\pi\sigma^2}}\right] \times e^{-\frac{(x - \mu)^2}{2\sigma^2}}$$

**Description:**
The **Probability Density Function (PDF) of a Gaussian (Normal) Distribution** with mean $\mu$ and standard deviation $\sigma$.
* **Domain / Parameters:** $x, \mu \in (-\infty, \infty)$ and scale parameter $\sigma > 0$.


```python
# Level 3 — Complex
# Question 13: [1 / √(2πσ²)] × e^[−(x − μ)² / (2σ²)]
import math

x = 1
mu = 0
sigma = 1
result = (1 / math.sqrt(2 * math.pi * sigma**2)) * math.exp(-((x - mu)**2) / (2 * sigma**2))

print("Expression: [1 / √(2πσ²)] × e^[−(x − μ)² / (2σ²)] [Gaussian PDF]")
print(f"Variables:  x = {x}, μ = {mu}, σ = {sigma}")
print(f"Result:     {result}")
```
