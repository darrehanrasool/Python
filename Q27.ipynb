## Question
Calculate depreciation using the reducing-balance method.

## Explanation
Enter initial asset value, annual depreciation rate as a percent, and years. The method applies the same rate to the remaining value each year.

## How to run
Run the code cell and enter the requested values when prompted.

```python
initial_value = float(input('Enter initial asset value: '))
depreciation_rate_percent = float(input('Enter annual depreciation rate (%): '))
years = float(input('Enter number of years: '))

## Reducing-balance value = initial value * (1 - rate)^years.
remaining_value = initial_value * (1 - depreciation_rate_percent / 100) ** years
total_depreciation = initial_value - remaining_value

print(f'Remaining value: {remaining_value:.2f}')
print(f'Total depreciation: {total_depreciation:.2f}')
```
