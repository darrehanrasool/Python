```python
## area of rectangle using python code 
## in python everthing is object.
## by default input funtion returns string type
length=int(input("enter the length buddy:"))
breadth=int(input("enter the breadth buddy:"))
print(type(length))
print(type(breadth))
## memory location address.
print(id(length))
print(id(breadth))
area=length*breadth
perimeter=2*(length+breadth)
print('area      of rectangle  = ',area)
print('perimeter of rectangle  = ',perimeter)
```
