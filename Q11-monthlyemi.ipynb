```python
principal = float(input('Enter loan principal: '))
annual_rate_percent = float(input('Enter annual interest rate (%): '))
time_years = float(input('Enter loan term in years: '))
monthly_rate = annual_rate_percent / 100 / 12
number_of_payments = time_years * 12

## EMI = P*r*(1+r)^n / ((1+r)^n - 1), with monthly rate r and payment count n.
## This standard formula assumes a positive monthly interest rate.
emi = principal * monthly_rate * (1 + monthly_rate) ** number_of_payments / ((1 + monthly_rate) ** number_of_payments - 1)

print(f'Monthly EMI: {emi:.2f}')
```
