```python
# Practical 5: File Handling - Write and read text files and calculate word frequency counts from file content.
import re

file_name = "lab_practical_05.txt"
content = '''Artificial intelligence and machine learning empower modern computing.
Python provides powerful tools for data science, artificial intelligence, and machine learning.
Learning Python is both practical and essential for artificial intelligence.'''

# Step 1: Write text to a file using context manager
with open(file_name, "w", encoding="utf-8") as f:
    f.write(content)
print(f"Successfully wrote data to '{file_name}'.")

# Step 2: Read text back from file
with open(file_name, "r", encoding="utf-8") as f:
    raw_text = f.read()

# Step 3: Count frequency of each word
words = re.findall(r'\b\w+\b', raw_text.lower())
word_frequency = {}
for word in words:
    word_frequency[word] = word_frequency.get(word, 0) + 1

# Display top frequent words
print(f"Total words found: {len(words)}")
print("Top word frequencies:")
for word, count in sorted(word_frequency.items(), key=lambda item: item[1], reverse=True)[:6]:
    print(f"  '{word}': {count}")
```
