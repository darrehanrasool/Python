## Question
Calculate a scientific expression involving several nested mathematical operations.

## Explanation
Example nested expression: sqrt(exp(x) + log(y)) / (1 + cos(z)^2). Enter x, positive y, and z in radians.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

x = float(input('Enter x: '))
y = float(input('Enter positive y: '))
z = float(input('Enter z (radians): '))

## Evaluate the inner exp/log/cos operations first, then the outer square root and division.
result = math.sqrt(math.exp(x) + math.log(y)) / (1 + math.cos(z) ** 2)

print(f'Nested expression result: {result:.6f}')
```
