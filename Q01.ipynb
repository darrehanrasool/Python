## Question
Calculate the area and circumference of a circle from its radius.

## Explanation
Enter the radius. The area measures the space inside the circle; the circumference is its boundary length.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

radius = float(input('Enter the circle radius: '))

## Circle formulas: area = pi * r^2; circumference = 2 * pi * r.
area = math.pi * radius ** 2
circumference = 2 * math.pi * radius

print(f'Area: {area:.2f}')
print(f'Circumference: {circumference:.2f}')
```
