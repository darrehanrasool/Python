```python
# Practical 3: Control Flow - Implement decision making with if-elif-else and iteration using for and while loops.
# Decision making with if, elif, and else
marks = 85
if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 50:
    grade = "B"
else:
    grade = "C"
print(f"Marks: {marks} -> Assigned Grade: {grade}")

# For loop: Iterating over sequences and ranges
print("For loop running sum:", end=" ")
total_sum = 0
for num in range(1, 6):
    total_sum += num
    print(f"+{num}", end=" ")
print(f"= {total_sum}")

# While loop: Repeating actions until condition is met
counter = 5
print("While loop countdown:", end=" ")
while counter > 0:
    print(counter, end=" ")
    counter -= 1
print("Liftoff!")
```
