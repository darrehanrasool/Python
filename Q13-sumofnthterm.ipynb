```python
first_term = float(input('Enter the first term: '))
common_difference = float(input('Enter the common difference: '))
n = int(input('Enter the number of terms: '))

## nth term: a + (n - 1)d. Sum: n/2 * (2a + (n - 1)d).
nth_term = first_term + (n - 1) * common_difference
progression_sum = n / 2 * (2 * first_term + (n - 1) * common_difference)

print(f'nth term: {nth_term:.2f}')
print(f'Sum of {n} terms: {progression_sum:.2f}')
```
