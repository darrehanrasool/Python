## Question
Calculate logarithm of a number with an arbitrary base.

## Explanation
Enter a positive number and a positive base other than 1. math.log(value, base) computes the logarithm in that base.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

value = float(input('Enter a positive number: '))
base = float(input('Enter a positive base (not 1): '))

## A logarithm answers: base raised to what power gives value?
logarithm = math.log(value, base)

print(f'Log base {base:g} of {value:g}: {logarithm:.6f}')
```
