```python
# Practical 15: Asynchronous Programming - Execute concurrent asynchronous tasks and measure performance using asyncio.
import asyncio
import time

async def download_resource(resource_id, duration):
    '''Simulates an asynchronous I/O operation with duration delay.'''
    print(f"-> [Start] Downloading Resource {resource_id} (estimated: {duration}s)")
    await asyncio.sleep(duration)
    print(f"<- [Done]  Resource {resource_id} download complete.")
    return f"Data_{resource_id}"

async def run_concurrent_pipeline():
    '''Runs multiple asynchronous tasks concurrently using asyncio.gather.'''
    start_time = time.perf_counter()
    tasks = [
        download_resource(1, 0.4),
        download_resource(2, 0.3),
        download_resource(3, 0.5)
    ]
    results = await asyncio.gather(*tasks)
    total_time = time.perf_counter() - start_time
    return results, total_time

# Execute the asynchronous pipeline
results, total_time = asyncio.run(run_concurrent_pipeline())
print(f"\nAll Tasks Completed: {results}")
print(f"Total Concurrent Execution Time: {total_time:.3f} seconds")
print("Observation: Tasks executed concurrently in parallel duration (~0.5s) rather than sequential sum (1.2s).")
```
