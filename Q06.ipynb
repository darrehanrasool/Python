## Question
Calculate the volume and surface area of a sphere.

## Explanation
Enter the sphere's radius. The formulas use pi and the radius.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

radius = float(input('Enter the sphere radius: '))

## Sphere volume = (4/3) * pi * r^3; surface area = 4 * pi * r^2.
volume = (4 / 3) * math.pi * radius ** 3
surface_area = 4 * math.pi * radius ** 2

print(f'Volume: {volume:.2f}')
print(f'Surface area: {surface_area:.2f}')
```
