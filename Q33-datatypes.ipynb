```python
# Question 33: Explore and demonstrate fundamental Python data types, memory sizes, literals, and number systems.

# 1. Integer Data Type & Memory Allocation in Python
import sys

age = 21                      # Small integer (28 bytes overhead in CPython)
salary = 85000                # Typical integer (28 bytes)
world_population = 8000000000  # Large integer (arbitrary precision: 32 bytes)

print('Integer values:', age, salary, world_population)
print('Data types:    ', type(age), type(salary), type(world_population))
print('Memory sizes:  ', sys.getsizeof(age), 'bytes,', sys.getsizeof(salary), 'bytes,', sys.getsizeof(world_population), 'bytes')
```

```python
# 2. Memory Locations & Object Identity using id()
# In Python, small integers (-5 to 256) share the same memory location (Integer Caching)
x = 100
y = 100
print('Value x =', x, 'Memory ID:', id(x))
print('Value y =', y, 'Memory ID:', id(y))
print('Do x and y reference the same memory?', id(x) == id(y))

# Reassigning creates a new integer object (integers are immutable)
x = x + 1
print('After increment: x =', x, 'New Memory ID:', id(x))
```

```python
# 3. Floating Point Representation (float)
# Standard decimals and scientific notation (E/e for powers of 10)
room_temp = 25.5              # Standard decimal float
speed_of_light = 3.0e8        # 3.0 * 10^8 m/s (scientific notation)
electron_charge = 1.6e-19     # 1.6 * 10^-19 C (negative exponent)

print('Standard float:      ', room_temp, 'Type:', type(room_temp))
print('Large scientific float:', speed_of_light, 'Type:', type(speed_of_light))
print('Small scientific float:', electron_charge, 'Type:', type(electron_charge))
```

```python
# 4. Boolean Data Type (bool)
# Evaluates to True or False; internally represented as integers (True=1, False=0)
is_registered = True
has_failed = False
is_eligible = (20 >= 18)       # Comparison expression  # noqa: PLR0133

print('is_registered:', is_registered, 'Type:', type(is_registered))
print('has_failed:   ', has_failed, 'Type:', type(has_failed))
print('is_eligible:  ', is_eligible, 'Type:', type(is_eligible))
print('Boolean arithmetic (True + True):', True + True)  # 1 + 1 = 2
```

```python
# 5. Complex Numbers (complex)
# In Python, 'j' or 'J' denotes the imaginary unit (j = sqrt(-1))
voltage = 230 + 15j           # AC circuit voltage
impedance = 50 - 20j          # AC circuit impedance

print('Voltage:   ', voltage, 'Type:', type(voltage))
print('Impedance: ', impedance, 'Type:', type(impedance))
print('Voltage Real Part:', voltage.real, '| Imaginary Part:', voltage.imag)
print('Sum of complex values:', voltage + impedance)
```

```python
# 6. Numeric Literals with Underscore Separators
# Underscores improve human readability for large numbers without affecting the value
annual_revenue = 1_000_000     # One Million
aadhaar_sample = 1234_5678_9012
credit_card = 4111_2222_3333_4444

print('Annual Revenue:   ', annual_revenue)
print('ID with separators:', aadhaar_sample)
print('Card Number:      ', credit_card)
```

```python
# 7. Number Systems (Decimal, Binary, Octal, Hexadecimal)
# Python natively supports 4 bases; all evaluate to their decimal values internally
decimal_num = 42              # Base 10
binary_num = 0b101010         # Base 2 (Prefix 0b or 0B)
octal_num = 0o52              # Base 8 (Prefix 0o or 0O)
hex_num = 0x2A                # Base 16 (Prefix 0x or 0X)
html_color = 0xFF5733         # Hexadecimal Web Color (#FF5733)

print('Decimal:    ', decimal_num, 'Type:', type(decimal_num))
print('Binary:     ', binary_num,  'Type:', type(binary_num))
print('Octal:      ', octal_num,   'Type:', type(octal_num))
print('Hexadecimal:', hex_num,     'Type:', type(hex_num))
print('HTML Color integer value:', html_color)
```
