```python
import math

x = float(input('Enter x (radians): '))
y = float(input('Enter positive y for log(y): '))
z = float(input('Enter nonnegative z for sqrt(z): '))
w = float(input('Enter w for exp(w): '))

## Example combines sine, cosine, natural logarithm, square root, and e raised to w.
result = math.sin(x) + math.cos(x) + math.log(y) + math.sqrt(z) + math.exp(w)

print(f'Expression result: {result:.6f}')
```
