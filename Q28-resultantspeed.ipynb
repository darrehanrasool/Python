```python
import math

vx = float(input('Enter horizontal velocity component (m/s): '))
vy = float(input('Enter vertical velocity component (m/s): '))

## Resultant speed = sqrt(vx^2 + vy^2); atan2 gives direction relative to +x.
speed = math.sqrt(vx ** 2 + vy ** 2)
direction_degrees = math.degrees(math.atan2(vy, vx))

print(f'Resultant velocity magnitude: {speed:.2f} m/s')
print(f'Direction: {direction_degrees:.2f} degrees')
```
