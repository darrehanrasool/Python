```python
## Numereic datatype in python
import sys
a=1
b=12
c=123456789123456789
print(a,b,c)
print('size =',sys.getsizeof(a),sys.getsizeof(b),sys.getsizeof(c))
```

```python
## Memory locations
x=4
print(id(x))
x=1234567891011
print(id(x))
```

```python
## Floating point representation
s=25.6
d=-3.5E-2
r=4.5e2
print(s,d,r)
```

```python
## Boolean 
a=True
b=False
print(type(a))
print(type(b))
a=1
b=2
print(type(a))
print(type(b))
```

```python
## Complex number
number_1=25+6j
number_2=13.5+6.5j
print(type(number_1))
print(type(number_2))
```

```python
## Literals
a=12345
b=12_34_5
print(b)
```

```python
## Number system
a=10
b=0b1010
c=0o12
d=0xA
print(a,b,c,d)
```
