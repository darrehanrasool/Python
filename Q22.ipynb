## Question
Calculate the angle between two vectors.

## Explanation
Enter the components of two 2D vectors. The cosine formula requires both vectors to be nonzero.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

ax = float(input('Vector A x component: '))
ay = float(input('Vector A y component: '))
bx = float(input('Vector B x component: '))
by = float(input('Vector B y component: '))

## cos(theta) = (A dot B) / (|A||B|); acos converts that ratio to an angle.
dot_product = ax * bx + ay * by
magnitude_a = math.sqrt(ax ** 2 + ay ** 2)
magnitude_b = math.sqrt(bx ** 2 + by ** 2)
cosine_angle = dot_product / (magnitude_a * magnitude_b)
angle_degrees = math.degrees(math.acos(cosine_angle))

print(f'Angle between vectors: {angle_degrees:.2f} degrees')
```
