```python
# Practical 10: Regular Expressions - Extract email addresses, phone numbers, and match patterns using the re module.
import re

sample_corpus = '''
Contact Information Directory:
- Admin Support: admin.support@iust.ac.in (Tel: +91-1933-247955)
- Dr. Rehan: rehan_ai@kashmir-univ.edu, alternate: rehan.researcher@gmail.com
- Admissions Office: admissions@ai-institute.org, emergency: info@tech-hub.net
- Invalid entries: test@.com, @domain.com, student@domain
'''

# Extract all valid email addresses
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
extracted_emails = re.findall(email_pattern, sample_corpus)

# Extract phone numbers
phone_pattern = r'\+?\d{1,3}[-.\s]?\d{3,4}[-.\s]?\d{4,6}'
extracted_phones = re.findall(phone_pattern, sample_corpus)

print(f"Extracted Emails ({len(extracted_emails)}):")
for email in extracted_emails:
    print(f"  - {email}")

print(f"\nExtracted Phone Numbers: {extracted_phones}")
```
