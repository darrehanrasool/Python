```python
principal = float(input('Enter principal amount: '))
annual_rate_percent = float(input('Enter annual interest rate (%): '))
periods_per_year = float(input('Enter compounding periods per year: '))
time_years = float(input('Enter time in years: '))

## A = P(1 + r/n)^(nt), where r is the annual rate as a decimal.
amount = principal * (1 + (annual_rate_percent / 100) / periods_per_year) ** (periods_per_year * time_years)
interest = amount - principal

print(f'Final amount: {amount:.2f}')
print(f'Compound interest: {interest:.2f}')
```
