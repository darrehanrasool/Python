```python
# Practical 16: Script Packaging and Distribution - Structure, configure, and package a Python script for distribution using pyproject.toml.
import os
import sys

package_root = "mathkit_pkg"
os.makedirs(f"{package_root}/src/mathkit", exist_ok=True)

# 1. Package Module Code
with open(f"{package_root}/src/mathkit/__init__.py", "w") as f:
    f.write('__version__ = "0.1.0"\nfrom .core import add, multiply\n')

with open(f"{package_root}/src/mathkit/core.py", "w") as f:
    f.write('''def add(a, b):
    """Returns the sum of two numbers."""
    return a + b

def multiply(a, b):
    """Returns the product of two numbers."""
    return a * b
''')

# 2. Modern pyproject.toml packaging configuration (PEP 621 / PEP 518)
with open(f"{package_root}/pyproject.toml", "w") as f:
    f.write('''[build-system]
requires = ["setuptools>=61.0"]
build-backend = "setuptools.build_meta"

[project]
name = "mathkit"
version = "0.1.0"
authors = [{ name="Dar Rehan", email="rehan@example.com" }]
description = "A lightweight mathematics utility library."
readme = "README.md"
requires-python = ">=3.8"
''')

# 3. Package documentation README
with open(f"{package_root}/README.md", "w") as f:
    f.write("# MathKit\nLightweight Python package demonstration for Lab 16.\n")

print(f"Created package directory tree at '{package_root}':")
for root, dirs, files in os.walk(package_root):
    level = root.replace(package_root, '').count(os.sep)
    indent = ' ' * 4 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = ' ' * 4 * (level + 1)
    for file in files:
        print(f"{subindent}{file}")

# 4. Test module import directly from source tree
sys.path.insert(0, os.path.abspath(f"{package_root}/src"))
import mathkit

print("\nImported Packaged Module:")
print(f"mathkit.__version__: {mathkit.__version__}")
print(f"mathkit.add(15, 27): {mathkit.add(15, 27)}")
print(f"mathkit.multiply(6, 7): {mathkit.multiply(6, 7)}")
```
