## Question
Convert a given number of seconds into hours, minutes, and seconds.

## Explanation
Enter the total seconds. Integer division and remainder separate full hours, minutes, and leftover seconds.

## How to run
Run the code cell and enter the requested values when prompted.

```python
total_seconds = int(input('Enter the number of seconds: '))

## Each hour has 3600 seconds. // finds whole units; % gives the remainder.
hours = total_seconds // 3600
remaining_seconds = total_seconds % 3600
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print(f'{hours} hour(s), {minutes} minute(s), {seconds} second(s)')
```
