```python
# Question 45: Find the maximum and minimum numbers from user inputs.
# maximum 
n=int(input("enter how many numbers total"))
i=0
# smallest possible number to compare
maxvalue=float('-inf')
while i<n:
    i+=1
    x=int(input(''))
    maxvalue = max(maxvalue, x)
print(f'maximum value : {maxvalue}')
```

```python
# minimum 
n=int(input("enter how many numbers total"))
i=0
# largest possible value to compare 
minvalue=float('inf')
while i<n:
    i+=1
    x=int(input(''))
    minvalue = min(minvalue, x)
print(f'minimum value : {minvalue}')
```

```python
# minimummaximum
n=int(input("enter how many numbers total"))
i=0
minvalue=float('inf')
maxvalue=float('-inf')
while i<n:
    i+=1
    x=int(input(''))
    minvalue = min(minvalue, x)
    maxvalue = max(maxvalue, x)
print(f'minimum value : {minvalue} and maximum value : {maxvalue}')
```
