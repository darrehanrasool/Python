```python
# maximum 
n=int(input("enter how many numbers total"))
i=0
# smallest possible number to compare
maxvalue=float('-inf')
while i<n:
    i+=1
    x=int(input(''))
    if x>maxvalue:
       maxvalue=x
print(f'maximum value : {maxvalue}')
```

```python
# minimum 
n=int(input("enter how many numbers total"))
i=0
# largest possible value to compare 
minvalue=float('inf')
while i<n:
    i+=1
    x=int(input(''))
    if x<minvalue:
       minvalue=x
print(f'minimum value : {minvalue}')
```

```python
# minimummaximum
n=int(input("enter how many numbers total"))
i=0
minvalue=float('inf')
maxvalue=float('-inf')
while i<n:
    i+=1
    x=int(input(''))
    if x<minvalue:
       minvalue=x
    if x>maxvalue:
       maxvalue=x
print(f'minimum value : {minvalue} and maximum value : {maxvalue}')
```
