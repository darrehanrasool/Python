```python
# maximum 
n=int(input("enter how many numbers total"))
i=0
maxvalue=0
while i<=n:
    i+=1
    x=int(input(''))
    if x>maxvalue:
       maxvalue=x
print(f'maximum value : {maxvalue}')
```

```python
# minimum 
n=int(input("enter how many numbers total"))
i=0
minvalue=1
while i<=n:
    i+=1
    x=int(input(''))
    if x<minvalue:
       minvalue=x
print(f'maximum value : {minvalue}')
```
