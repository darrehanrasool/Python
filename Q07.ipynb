## Question
Calculate the average of three numbers.

## Explanation
Enter three numbers. Their arithmetic mean is their sum divided by three.

## How to run
Run the code cell and enter the requested values when prompted.

```python
first = float(input('Enter the first number: '))
second = float(input('Enter the second number: '))
third = float(input('Enter the third number: '))

## Average = sum of values divided by the number of values.
average = (first + second + third) / 3

print(f'Average: {average:.2f}')
```
