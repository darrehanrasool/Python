## Question
Calculate BMI from weight and height.

## Explanation
Enter weight in kilograms and height in metres. BMI is a ratio of weight to squared height.

## How to run
Run the code cell and enter the requested values when prompted.

```python
weight_kg = float(input('Enter weight in kilograms: '))
height_m = float(input('Enter height in metres: '))

## BMI formula: weight in kilograms divided by height in metres squared.
bmi = weight_kg / height_m ** 2

print(f'BMI: {bmi:.2f}')
```
