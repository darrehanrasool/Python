## Question
Calculate the resultant of two forces.

## Explanation
Enter force magnitudes and the angle between them in degrees. The direction is measured from force 1 toward force 2.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

force1 = float(input('Enter force 1 magnitude: '))
force2 = float(input('Enter force 2 magnitude: '))
angle_degrees = float(input('Enter angle between forces (degrees): '))
angle_radians = math.radians(angle_degrees)

## By the cosine rule, resultant magnitude = sqrt(F1^2 + F2^2 + 2*F1*F2*cos(theta)).
resultant = math.sqrt(force1 ** 2 + force2 ** 2 + 2 * force1 * force2 * math.cos(angle_radians))

print(f'Resultant force magnitude: {resultant:.2f}')
```
