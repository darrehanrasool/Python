```python
# Practical 2: Data Types and Variables - Create variables of fundamental types and perform core arithmetic, string, and collection operations.
# Numeric data types and basic arithmetic
num_int = 42
num_float = 3.14159
num_sum = num_int + num_float
print(f"Integer: {num_int} ({type(num_int).__name__})")
print(f"Float: {num_float} ({type(num_float).__name__})")
print(f"Sum: {num_sum:.4f}")

# String manipulation
first_word = "Python"
second_word = "Programming"
combined_str = first_word + " " + second_word
joined_str = f"{first_word}-Lab-2026"
print(f"Concatenated String: '{combined_str}'")
print(f"Joined String: '{joined_str}'")

# List operations
items_list = [10, 20, 30]
items_list.append(40)
print(f"List after append: {items_list} (Length: {len(items_list)})")

# Tuple operations
dim_tuple = (1920, 1080)
width, height = dim_tuple
print(f"Unpacked Tuple: Width={width}, Height={height}")

# Dictionary operations
user_dict = {"name": "Rehan", "role": "Researcher", "semester": 1}
user_dict["credits"] = 2
print(f"Dictionary keys: {list(user_dict.keys())}")
print(f"Dictionary contents: {user_dict}")
```
