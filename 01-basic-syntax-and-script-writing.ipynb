```python
# Practical 1: Basic Syntax and Script Writing - Take user input, calculate the area and circumference of a circle, and output formatted results.
import math

try:
    user_val = input("Enter the circle radius: ")
    radius = float(user_val) if user_val.strip() else 7.0
except (ValueError, EOFError):
    radius = 7.0

# Calculate circle metrics: area = pi * r^2, circumference = 2 * pi * r
area = math.pi * (radius ** 2)
circumference = 2 * math.pi * radius

print(f"Radius: {radius:.2f}")
print(f"Area: {area:.2f}")
print(f"Circumference: {circumference:.2f}")
```
