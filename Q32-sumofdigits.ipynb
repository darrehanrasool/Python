```python
number = int(input("Enter a three-digit positive integer: "))

## // 100 gives the hundreds digit.
hundreds_digit = number // 100

## // 10 removes the ones digit; % 10 gives the tens digit.
tens_digit = (number // 10) % 10

## % 10 gives the ones digit.
ones_digit = number % 10

## Add all three digits.
digit_sum = hundreds_digit + tens_digit + ones_digit

## Display the digits and their sum.
print(f"Digits: {hundreds_digit}, {tens_digit}, {ones_digit}")
print(f"Sum of digits: {digit_sum}")
```

```python
number = int(input("Enter a four-digit positive integer: "))

## // 1000 gives the thousands digit.
thousands_digit = number // 1000

## // 100 removes the last two digits; % 10 gives the hundreds digit.
hundreds_digit = (number // 100) % 10

## // 10 removes the ones digit; % 10 gives the tens digit.
tens_digit = (number // 10) % 10

## % 10 gives the ones digit.
ones_digit = number % 10

## Add all four digits.
digit_sum = thousands_digit + hundreds_digit + tens_digit + ones_digit

## Display the digits and their sum.
print(f"Digits: {thousands_digit}, {hundreds_digit}, {tens_digit}, {ones_digit}")
print(f"Sum of digits: {digit_sum}")
```

```python
number = int(input("Enter a positive integer: "))

## Store the original number for displaying it later.
original_number = number

## Start the digit sum at 0.
digit_sum = 0

## Repeat until all digits have been processed.
while number > 0:

    ## % 10 extracts the last digit.
    digit = number % 10

    ## Add the extracted digit to the sum.
    digit_sum = digit_sum + digit

    ## // 10 removes the last digit.
    number = number // 10

## Display the result.
print(f"Number: {original_number}")
print(f"Sum of digits: {digit_sum}")
```
