```python
number = int(input('Enter a four-digit positive integer: '))

## // 1000 removes the last three digits, leaving the thousands digit.
thousands_digit = number // 1000

## // 100 removes the last two digits; % 10 selects the hundreds digit.
hundreds_digit = (number // 100) % 10

## // 10 removes the ones digit; % 10 selects the tens digit.
tens_digit = (number // 10) % 10

## % 10 extracts the rightmost, ones digit.
ones_digit = number % 10

## Add all four extracted digits.
digit_sum = thousands_digit + hundreds_digit + tens_digit + ones_digit

## Display the digits and their sum.
print(f'Digits: {thousands_digit}, {hundreds_digit}, {tens_digit}, {ones_digit}')
print(f'Sum of digits: {digit_sum}')
```
