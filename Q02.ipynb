## Question
Convert Celsius temperature to Fahrenheit and Kelvin.

## Explanation
Enter a temperature in Celsius. The formulas convert it to the Fahrenheit and Kelvin scales.

## How to run
Run the code cell and enter the requested values when prompted.

```python
celsius = float(input('Enter temperature in Celsius: '))

## Fahrenheit scales Celsius by 9/5 and adds 32; Kelvin adds 273.15.
fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

print(f'Fahrenheit: {fahrenheit:.2f} °F')
print(f'Kelvin: {kelvin:.2f} K')
```
