```python
x1 = float(input('Enter x1: '))
y1 = float(input('Enter y1: '))
x2 = float(input('Enter x2: '))
y2 = float(input('Enter y2: '))

## Midpoint coordinates are the averages of x values and y values.
midpoint_x = (x1 + x2) / 2
midpoint_y = (y1 + y2) / 2

print(f'Midpoint: ({midpoint_x:.2f}, {midpoint_y:.2f})')
```
