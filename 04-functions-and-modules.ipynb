```python
# Practical 4: Functions and Modules - Define custom modular functions and leverage Python standard library modules.
import math
import random
import datetime

# Custom function definition with default arguments
def calculate_cylinder_volume(radius, height=10.0):
    '''Calculates the volume of a cylinder given radius and height.'''
    volume = math.pi * (radius ** 2) * height
    return volume

# Function demonstrating variable scope
base_scale = 1.5

def scaled_computation(val):
    local_factor = 2.0
    return val * local_factor * base_scale

# Standard library modules in action
cyl_vol = calculate_cylinder_volume(3.0, 5.0)
scaled_res = scaled_computation(4.0)
rand_sample = random.sample(range(10, 100), 3)
current_date = datetime.date.today().isoformat()

print(f"Cylinder Volume (r=3, h=5): {cyl_vol:.2f}")
print(f"Scaled Computation Result: {scaled_res:.2f}")
print(f"Standard Library Random Sample: {rand_sample}")
print(f"Standard Library Current Date: {current_date}")
```
