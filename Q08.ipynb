## Question
Calculate the distance between two points.

## Explanation
Enter x and y coordinates for both points. The distance formula comes from the Pythagorean theorem.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

x1 = float(input('Enter x1: '))
y1 = float(input('Enter y1: '))
x2 = float(input('Enter x2: '))
y2 = float(input('Enter y2: '))

## Distance = square root of the sum of squared coordinate differences.
distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print(f'Distance: {distance:.2f}')
```
