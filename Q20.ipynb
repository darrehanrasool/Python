## Question
Calculate the range of a projectile.

## Explanation
Enter launch speed in metres per second and angle in degrees. Assume equal launch and landing heights and ignore air resistance.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

initial_speed = float(input('Enter launch speed (m/s): '))
angle_degrees = float(input('Enter launch angle (degrees): '))
gravity = 9.81
angle_radians = math.radians(angle_degrees)

## For equal heights, range = u^2*sin(2*theta)/g.
projectile_range = initial_speed ** 2 * math.sin(2 * angle_radians) / gravity

print(f'Horizontal range: {projectile_range:.2f} metres')
```
