```python
import math

first = float(input('Enter the first number: '))
second = float(input('Enter the second number: '))
third = float(input('Enter the third number: '))

## Population standard deviation = sqrt(mean of squared differences from the mean).
mean = (first + second + third) / 3
variance = ((first - mean) ** 2 + (second - mean) ** 2 + (third - mean) ** 2) / 3
standard_deviation = math.sqrt(variance)

print(f'Mean: {mean:.2f}')
print(f'Population standard deviation: {standard_deviation:.2f}')
```
