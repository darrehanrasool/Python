```python
import math

radius = float(input('Enter radius: '))
angle_degrees = float(input('Enter angle in degrees: '))
angle_radians = math.radians(angle_degrees)

## x = r*cos(theta), y = r*sin(theta); math trig functions use radians.
## applying trignometeric funtions angle must be in radians
x = radius * math.cos(angle_radians)
y = radius * math.sin(angle_radians)

print(f'Cartesian coordinates: ({x:.2f}, {y:.2f})')
```
