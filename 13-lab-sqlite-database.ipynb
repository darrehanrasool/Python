```python
# Practical 13: Database Interaction - Connect to a SQLite database and execute Create, Read, Update, and Delete (CRUD) operations using sqlite3.
import sqlite3

# Connect to database (in-memory or local file)
conn = sqlite3.connect("lab_students.db")
cursor = conn.cursor()

# 1. CREATE Table
cursor.execute('''
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    department TEXT NOT NULL,
    gpa REAL NOT NULL
)
''')

# 2. INSERT (Create records)
cursor.execute("DELETE FROM students")  # Clean state for repeatable runs
sample_students = [
    ("Dar Rehan", "Artificial Intelligence", 3.92),
    ("Ahmad Malik", "Computer Science", 3.65),
    ("Zubair Lone", "Data Engineering", 3.40)
]
cursor.executemany("INSERT INTO students (name, department, gpa) VALUES (?, ?, ?)", sample_students)
conn.commit()
print("Inserted 3 student records.")

# 3. READ (Select records)
cursor.execute("SELECT id, name, department, gpa FROM students")
print("Initial Records:")
for row in cursor.fetchall():
    print(f"  ID {row[0]}: {row[1]} | {row[2]} | GPA: {row[3]:.2f}")

# 4. UPDATE Record
cursor.execute("UPDATE students SET gpa = ? WHERE name = ?", (3.98, "Dar Rehan"))
conn.commit()
print("\nUpdated Dar Rehan's GPA to 3.98.")

# 5. DELETE Record
cursor.execute("DELETE FROM students WHERE name = ?", ("Zubair Lone",))
conn.commit()
print("Deleted Zubair Lone's record.")

# Verify final state
cursor.execute("SELECT id, name, department, gpa FROM students")
print("\nFinal Records in Database:")
for row in cursor.fetchall():
    print(f"  ID {row[0]}: {row[1]} | {row[2]} | GPA: {row[3]:.2f}")

conn.close()
```
