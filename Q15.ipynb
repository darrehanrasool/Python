## Question
Convert Cartesian coordinates to polar coordinates.

## Explanation
Enter Cartesian coordinates x and y. The angle is reported in radians and degrees.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

x = float(input('Enter x: '))
y = float(input('Enter y: '))

## Polar radius r = sqrt(x^2 + y^2); atan2(y, x) finds the angle in the correct quadrant.
radius = math.sqrt(x ** 2 + y ** 2)
angle_radians = math.atan2(y, x)
angle_degrees = math.degrees(angle_radians)

print(f'Polar coordinates: (r={radius:.2f}, theta={angle_radians:.4f} radians / {angle_degrees:.2f} degrees)')
```
