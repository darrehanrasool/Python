## Question
Calculate the coordinates of a point after rotation by a given angle.

## Explanation
Enter point coordinates and a counterclockwise rotation angle in degrees about the origin.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

x = float(input('Enter point x: '))
y = float(input('Enter point y: '))
angle_degrees = float(input('Enter counterclockwise rotation angle (degrees): '))
theta = math.radians(angle_degrees)

## Rotation: x' = x*cos(theta) - y*sin(theta); y' = x*sin(theta) + y*cos(theta).
rotated_x = x * math.cos(theta) - y * math.sin(theta)
rotated_y = x * math.sin(theta) + y * math.cos(theta)

print(f'Rotated point: ({rotated_x:.2f}, {rotated_y:.2f})')
```
