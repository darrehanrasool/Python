## Question
Calculate the time of flight of a projectile.

## Explanation
Enter launch speed in metres per second and launch angle in degrees. Assume launch and landing heights are equal and ignore air resistance.

## How to run
Run the code cell and enter the requested values when prompted.

```python
import math

initial_speed = float(input('Enter launch speed (m/s): '))
angle_degrees = float(input('Enter launch angle (degrees): '))
gravity = 9.81
angle_radians = math.radians(angle_degrees)

## Vertical launch speed is u*sin(theta); flight time for equal heights is 2*u_y/g.
time_of_flight = 2 * initial_speed * math.sin(angle_radians) / gravity

print(f'Time of flight: {time_of_flight:.2f} seconds')
```
