```python
number = int(input('Enter a three-digit positive integer: '))

## // 100 drops the last two digits, leaving the hundreds digit.
hundreds_digit = number // 100

## % 10 gives the remainder after division by 10, which is the ones digit.
ones_digit = number % 10

## // 10 drops the ones digit; % 10 then extracts the tens digit.
tens_digit = (number // 10) % 10

## Add the three extracted digits to get their sum.
digit_sum = hundreds_digit + tens_digit + ones_digit

## Display the digits and their sum.
print(f'Digits: {hundreds_digit}, {tens_digit}, {ones_digit}')
print(f'Sum of digits: {digit_sum}')
```
