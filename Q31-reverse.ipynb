```python
number = int(input('Enter a three-digit positive integer: '))

## % 10 extracts the original ones digit; it becomes the reversed hundreds digit.
ones_digit = number % 10

## // 10 removes the original ones digit. % 10 then extracts the original tens digit.
tens_digit = (number // 10) % 10

## // 100 removes the last two digits, leaving the original hundreds digit.
hundreds_digit = number // 100

## Rebuild the number in reverse place order: ones*100 + tens*10 + hundreds.
reversed_number = ones_digit * 100 + tens_digit * 10 + hundreds_digit

## Display the reversed number.
print(f'Reversed number: {reversed_number}')
```
