```python
# Practical 7: Classes and Object-Oriented Programming - Define a Book class with attributes and methods to manage library book records.
class Book:
    def __init__(self, title, author, isbn, pages):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.pages = pages
        self.is_borrowed = False

    def display_info(self):
        status = "Borrowed" if self.is_borrowed else "Available"
        print(f"Book: '{self.title}' | Author: {self.author} | Pages: {self.pages} | Status: {status}")

    def borrow_book(self):
        if not self.is_borrowed:
            self.is_borrowed = True
            print(f"-> Checked out '{self.title}'.")
        else:
            print(f"-> '{self.title}' is currently unavailable.")

    def return_book(self):
        if self.is_borrowed:
            self.is_borrowed = False
            print(f"-> Returned '{self.title}'.")
        else:
            print(f"-> '{self.title}' was not borrowed.")

    def __str__(self):
        return f"Book('{self.title}' by {self.author})"

# Demonstration of class instantiation and method calls
book1 = Book("Deep Learning", "Ian Goodfellow", "978-0262035613", 800)
book2 = Book("Pattern Recognition and Machine Learning", "Christopher Bishop", "978-0387310732", 738)

print(book1)
book1.display_info()
book1.borrow_book()
book1.display_info()
book1.return_book()
```
