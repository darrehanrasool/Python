```python
import math

initial_speed = float(input('Enter launch speed (m/s): '))
angle_degrees = float(input('Enter launch angle (degrees): '))
gravity = 9.81
angle_radians = math.radians(angle_degrees)

## The peak height is vertical speed squared divided by 2g.
vertical_speed = initial_speed * math.sin(angle_radians)
maximum_height = vertical_speed ** 2 / (2 * gravity)

print(f'Maximum height: {maximum_height:.2f} metres')
```
