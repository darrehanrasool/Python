```python
# Unit 1: Python Fundamentals - Development environments, syntax, variables, data types, control structures, functions, modules, data structures, and file handling.
# 1. Development environments and basic syntax
# Jupyter Notebooks provide interactive execution; IDEs like VS Code and PyCharm offer comprehensive debugging and project management.
app_name = "Python Environment Demo"
version = 3.14
is_interactive = True

print(f"Environment: {app_name} | Version: {version} | Interactive: {is_interactive}")
print("Variable Types:", type(app_name), type(version), type(is_interactive))
```

```python
# 2. Control structures: Decision making (if-elif-else) and iteration (for, while)
score = 88
if score >= 90:
    grade = "A+"
elif score >= 75:
    grade = "A"
elif score >= 50:
    grade = "B"
else:
    grade = "F"
print(f"Score: {score} -> Grade: {grade}")

# For loop with range
even_squares = [i**2 for i in range(1, 6) if i % 2 == 0]
print("Even squares (1 to 5):", even_squares)

# While loop with condition
countdown = 3
print("Countdown:", end=" ")
while countdown > 0:
    print(countdown, end=" ")
    countdown -= 1
print("Done!")
```

```python
# 3. Functions, scope, and importing standard library modules
import math

global_multiplier = 2

def calculate_circle_metrics(radius, scale=1.0):
    '''Computes circumference and area with scaling.'''
    effective_radius = radius * scale * global_multiplier
    circ = 2 * math.pi * effective_radius
    area = math.pi * (effective_radius ** 2)
    return circ, area

circ, area = calculate_circle_metrics(3.5)
print(f"Global Multiplier: {global_multiplier}")
print(f"Calculated Circumference: {circ:.2f}, Area: {area:.2f}")
```

```python
# 4. Built-in data structures and comprehensions
# Lists, tuples, sets, and dictionaries
fruits = ["apple", "banana", "cherry"]
fruits.append("orange")

point = (10, 20)

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
union_set = set_a | set_b
intersection_set = set_a & set_b

student_info = {"id": 101, "name": "Dar Rehan", "course": "M.Tech AI"}

# List and dictionary comprehensions
squares_comp = [x**2 for x in range(1, 6)]
char_freq = {char: "AI-Kashmir".count(char) for char in set("AI-Kashmir")}

print("Fruits list:", fruits)
print("Coordinate tuple:", point)
print("Union:", union_set, "| Intersection:", intersection_set)
print("Student dict:", student_info)
print("Squares comprehension:", squares_comp)
print("Char count comprehension:", char_freq)
```

```python
# 5. File handling: Safe reading and writing using context managers
demo_filename = "unit1_demo_file.txt"

# Write data to file
with open(demo_filename, "w", encoding="utf-8") as file:
    file.write("Python Unit 1: Foundations")
    file.write("File handling demonstration.")

# Read and parse file
with open(demo_filename, "r", encoding="utf-8") as file:
    lines = file.readlines()

print(f"Read {len(lines)} lines from '{demo_filename}':")
for idx, line in enumerate(lines, start=1):
    print(f"  Line {idx}: {line.strip()}")
```
