```python
import math

x = float(input('Enter x: '))
y = float(input('Enter positive y: '))
z = float(input('Enter z (radians): '))

## Evaluate the inner exp/log/cos operations first, then the outer square root and division.
result = math.sqrt(math.exp(x) + math.log(y)) / (1 + math.cos(z) ** 2)

print(f'Nested expression result: {result:.6f}')
```
