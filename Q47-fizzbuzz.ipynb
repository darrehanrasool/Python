```python
n = int(input("Enter the number till I need to get FizzBuzz: "))

i = 1

while i <= n:
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    elif i % 3 != 0 and i % 5 != 0:
        i += 1
        continue

    i += 1
```
