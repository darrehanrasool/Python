## Question
Calculate simple interest and total amount.

## Explanation
Enter principal, annual interest rate as a percent, and time in years.

## How to run
Run the code cell and enter the requested values when prompted.

```python
principal = float(input('Enter principal amount: '))
annual_rate_percent = float(input('Enter annual interest rate (%): '))
time_years = float(input('Enter time in years: '))

## Simple interest = principal * rate(decimal) * time.
interest = principal * (annual_rate_percent / 100) * time_years
total_amount = principal + interest

print(f'Simple interest: {interest:.2f}')
print(f'Total amount: {total_amount:.2f}')
```
