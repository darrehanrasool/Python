```python
# Question 40: Demonstrate explicit type conversion (typecasting) across int, float, bool, complex, and str.

# 1. Integer Conversion using int()
# Converts float (truncates decimal), bool, and valid numeric strings into integers
float_val = 99.85
bool_t = True
bool_f = False
str_num = '450'
bin_str = '1010'      # Base 2 string
hex_str = 'FF'        # Base 16 string

print('float 99.85 -> int:     ', int(float_val))        # Truncates to 99
print('bool True -> int:        ', int(bool_t))           # 1
print('bool False -> int:       ', int(bool_f))           # 0
print('str "450" -> int:        ', int(str_num))          # 450
print('binary str "1010" -> int:', int(bin_str, 2))      # 10
print('hex str "FF" -> int:     ', int(hex_str, 16))     # 255
```

```python
# 2. Floating Point Conversion using float()
# Converts integers, booleans, and valid decimal/scientific strings into floats
int_val = 75
bool_t = True
bool_f = False
dec_str = '89.75'
sci_str = '1.5e3'     # 1.5 * 10^3 = 1500.0
bin_literal = 0b1010  # Binary literal (10)

print('int 75 -> float:       ', float(int_val))
print('bool True -> float:    ', float(bool_t))
print('bool False -> float:   ', float(bool_f))
print('str "89.75" -> float:  ', float(dec_str))
print('str "1.5e3" -> float:  ', float(sci_str))
print('binary 0b1010 -> float:', float(bin_literal))
```

```python
# 3. Boolean Conversion using bool()
# Falsy values evaluate to False; all other values evaluate to True
# Falsy values in Python: 0, 0.0, '', None, False

print('--- FALSY EXAMPLES (evaluate to False) ---')
print('bool(0):       ', bool(0))
print('bool(0.0):     ', bool(0.0))
print('bool(""):       ', bool(''))       # Empty string
print('bool(None):    ', bool(None))
print('bool(False):   ', False)

print('\n--- TRUTHY EXAMPLES (evaluate to True) ---')
print('bool(42):      ', bool(42))      # Non-zero integer
print('bool(-15):     ', bool(-15))     # Negative non-zero integer
print('bool("Hello"): ', bool('Hello')) # Non-empty string
print('bool("False"): ', bool('False')) # Non-empty string (even word False is True!)
print('bool("0"):     ', bool('0'))     # Non-empty string (contains character '0')
```

```python
# 4. Complex Number Conversion using complex()
# Forms complex number real + imag*j
int_val = 12
float_val = -8.5
bool_t = True
str_complex = '3.5+4.5j'

print('int 12 -> complex:          ', complex(int_val))
print('float -8.5 -> complex:      ', complex(float_val))
print('bool True -> complex:       ', complex(bool_t))
print('complex(real=5, imag=12):   ', complex(5, 12))
print('str "3.5+4.5j" -> complex:  ', complex(str_complex))
```

```python
# 5. String Conversion using str()
# Converts any Python object into its readable string representation
score = 95
temperature = 37.2
is_valid = True
equation = 3 + 4j

print('int -> str:    ', repr(str(score)), 'Type:', type(str(score)))
print('float -> str:  ', repr(str(temperature)), 'Type:', type(str(temperature)))
print('bool -> str:   ', repr(str(is_valid)), 'Type:', type(str(is_valid)))
print('complex -> str:', repr(str(equation)), 'Type:', type(str(equation)))

# Practical usage: string concatenation
print('Student final score message: ' + str(score) + '/100')
```
